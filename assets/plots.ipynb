# Underwater Cables Empirical Results

Run this notebook to visualize the benchmark measurements written by `src/benchmark.py`.

```python
import csv
from pathlib import Path

import matplotlib.pyplot as plt

csv_path = next(
    path for path in (
        Path("empirical_results.csv"),
        Path("assets/empirical_results.csv"),
    )
    if path.exists()
)
with csv_path.open("r", newline="", encoding="utf-8") as csv_file:
    results = list(csv.DictReader(csv_file))

vertices = [int(row["vertices"]) for row in results]
edge_counts = [int(row["edges"]) for row in results]
times = [float(row["average_runtime_ms"]) for row in results]
costs = [int(row["mst_total_cost"]) for row in results]
```

```python
plt.figure(figsize=(8, 5))
plt.plot(vertices, times, marker="o")
plt.xlabel("Number of vertices")
plt.ylabel("Average runtime (milliseconds)")
plt.title("Underwater Cables Benchmark Runtime")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(vertices, edge_counts, marker="o")
plt.xlabel("Number of vertices")
plt.ylabel("Number of edges")
plt.title("Generated Graph Size")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(vertices, costs, marker="o", color="green")
plt.xlabel("Number of vertices")
plt.ylabel("Minimum spanning tree cost")
plt.title("Minimum Cable Cost")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
